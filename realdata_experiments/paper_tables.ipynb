```python
from tables import build_df, latex_main_table, latex_appendix, compact_volumes

EXPERIMENTS = {
    'Bio':   {2: 'results_clean/bio_2d/*.json',
              3: 'results_clean/bio_3d/*.json',
              4: 'results_clean/bio_4d/*.json'},
    'House': {2: 'results_clean/house_2d/*.json',
              3: 'results_clean/house_3d/*.json',
              4: 'results_clean/house_4d/*.json',
              6: 'results_clean/house_6d/*.json',
              8: 'results_clean/house_8d/*.json',
              10: 'results_clean/house_10d/*.json'},
    'Blog':  {2: 'results_clean/blog_2d/*.json',
              3: 'results_clean/blog_3d/*.json',
              4: 'results_clean/blog_4d/*.json'},
}

df = build_df(EXPERIMENTS)
print(df.groupby(['dataset', 'd', 'model']).seed.nunique().to_string())
```

```python
# Table 2
print(latex_main_table(df, 'Ridge', dataset_order=('Bio', 'Blog', 'House'),
                       dims={'Bio': (2, 3, 4), 'Blog': (2, 3, 4), 'House': (2, 3, 4, 6, 8, 10)}))
```

```python
# Appendix tables
for ds, dims in EXPERIMENTS.items():
    for d in sorted(dims):
        n = df[(df.dataset == ds) & (df.d == d)].groupby('model').seed.nunique().min()
        tex = latex_appendix(df, ds, d,
            caption=f"{ds} dataset ($d={d}$), Ridge and MLP. Mean $\\pm$ std over {n} seeds.",
            label=f"tab:{ds.lower()}_{d}d")
        print(compact_volumes(tex))
        print('\n' * 3)
```
